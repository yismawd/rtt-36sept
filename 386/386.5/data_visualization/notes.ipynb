```python
# matplotlib function
plt.plot() - default creates line chart (trends over time)
plt.show() - start an eventloop, that opens interactive window to view active figures (figs) similar to print but for plots
plt.scatter() - scater plot, visualizing relationships
plt.bar() - comparing categorical or discrete data
plt.barh() - hz barplot
plt.hist() - barplot for continuous data/quantitative data 
plt.savefig() - can save a plot to a file 
plt.contour() - draws contour lines
plt.annotate() - create annotation on your plot
```

Plot vocabulary: 
-   figure (fig) - container, window to put your plots on
    - figsize - size of the figure in inches
    - dpi - dots per inch (resolution)
    - facecolor - background color
    - subplotspars - distance between subplots
-   axes (ax) - area within the fig where data is plotted
    - title - title for the chart/plot
    - xlable, ylable, xscale, yscale, xlim, ylim
-   axis - x, y axis of a plot or chart
-   line - data itself, cant be fully customized
    - color, linestyle, linewidth, markers (symbols that represent data points), markersize
-   Legend - small container with labels/descriptors of elements
-   labels - names/labels for the x and y axis
-   title - title for the chart/plot
-   markers - symbols that represent data points
-   grid - backgrod (attribute that can be edited)
-   ticks (tic) - marks along the x,y axis to represent values

# approaches to creating plots
1 - state based approach - implicitly create and manipulate figures (imlicit - no control) - less upfront code, but less control
2 - object oriented approach - explicitly creating and namuplating your axes and figures - more upfront code, but more control

```python
# state based 
x_points = np.array([1, 13])
y_points = np.array([2, 12])
plt.plot(x_points, y_points)
plt.show()
```

```python
# object oreinted approach - you can change the size, type, color, ...
fig, ax = plt.subplots(figsize =(2,2))
ax.plot(x_points, y_points, linestyle='--', color='red')
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])

# Create a lineplot, custom cuolor, dotted linestyle, diamond markers
plt.figure(figsize=(6,2), dpi=144)
plt.plot(
    x, y,
    linestyle=':',
    marker='D',
    color=(0.5, 0.5, 0.8)
)

plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')
plt.grid(color='red', linestyle='--', alpha=0.2)
plt.gca().set_facecolor('lightblue')
plt.show()
```

```python
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])
plt.subplot(1, 2, 1) # create 1 row , with 2 plots, this plot is index 1
plt.plot(x, y)
plt.title('Plot 1')

x1 = np.array([0, 1, 2, 3])
y1 = np.array([10, 20, 25, 30])
plt.subplot(1, 2, 2) # create 1 row, with 2 plots, this is index 2
plt.title('Plot 2')
plt.plot(x1, y1)
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y1 = np.array([3, 4, 5, 7, 11])
y2 = np.array([1, 4, 9, 16, 25])

fig, axs = plt.subplots(2, 1)
axs[0].plot(x, y1)
axs[0].set_title('Plot 1')

axs[1].plot(x, y2)
axs[1].set_title('Plot 2')

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(x)
```

```python
fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y1)
axes[0, 0].set_title('Sin Function')

axes[0, 1].plot(x, y2)
axes[0, 1].set_title('Cosine Function')

axes[1, 0].plot(x, y3)
axes[1, 0].set_title('Tangent Function')

axes[1, 1].plot(x, y4)
axes[1, 1].set_title('Exponential Function')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 1)

axes[0].plot(x, y1)
axes[0].set_title('Sin Function')

axes[1].plot(x, y2)
axes[1].set_title('Cosine Function')

axes[2].plot(x, y3)
axes[2].set_title('Tangent Function')

axes[3].plot(x, y4)
axes[3].set_title('Exponential Function')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 1, sharex=True, )

axes[0].plot(x, y1)
axes[0].set_title('Sin Function')

axes[1].plot(x, y2)
axes[1].set_title('Cosine Function')

axes[2].plot(x, y3)
axes[2].set_title('Tangent Function')

axes[3].plot(x, y4)
axes[3].set_title('Exponential Function')

plt.tight_layout()
plt.show()
```

```python
years=[1950,1955,1960,1965,1970,1980,1985,1990,1995,2000,2005,2010,2015]
pops=[2.5,2.7,3.0,3.3,3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3]
death=[1.2,1.1,1.2,2.1,2.0,2.3,1.8,1.9,2.6,1.6,2.4,2.4,4.0]
job_rate = [0.5, 0.6, 0.8, 0.9, 1.0, 1.2, 1.5, 1.7, 2.0, 2.2, 2.5, 2.7, 2.9]
```

```python
plt.plot(years, pops, color='red', linestyle='--')
plt.plot(years, death, color='green', linestyle=':')
plt.title('Population Growth')
plt.xlable('Population Growth by Year')
plt.ylable('Population (Billions)')
plt.show()
```

```python
plt.plot(years, pops, color='red', linestyle='--', label='Population')
plt.plot(years, death, color='green', linestyle=':', label='Deaths')

plt.title('Population Growth')
plt.xlabel('Population Growth by Year')
plt.ylabel('Population (billions)')
plt.legend()
plt.grid(alpha=0.3, linestyle='--')
plt.show()
```

```python
plt.plot(years, pops, color='red', linestyle='--', label='Population')
plt.plot(years, death, color='green', linestyle=':', label='Death')
plt.plot(years, job_rate, color='purple', linestyle='-.', label='Job Rate')

plt.title('Population Growth')
plt.xlabel('Population Growth by Year')
plt.ylabel('Population (Billions)')

plt.legend() # when using a legend. the plots must have labels. See .plot() above
plt.grid(alpha=0.3, linestyle='--')
plt.show()
```
