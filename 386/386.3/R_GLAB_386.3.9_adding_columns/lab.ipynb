R_GLAB 386.3.9 - adding columns to DataFrames

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Name': ['Jane', 'Princi', 'James', 'Fadi'],
    'Height': [5.1, 6.2, 5.1, 5.2],
    'Qualifications': ['Msc', 'MA', 'Msc', 'Msc'],
    'Score 1': [50, 86, 77, 45],
    'Score 2': [50, 96, 60, 30]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
address = ['NYC', 'NJ', 'PA', 'CA']
df['address'] = address
print(df)
```

```python
# create a new total score 
df['Tatal Score'] = df['Score 1'] + df['Score 2']
print(df)
```

```python
# create an average score column
df['Average Score'] = (df['Score 1'] + df['Score 2']) / 2
print(df)
```

```python
# creating and initializing a list
values = [['Rohan', 455], ['Elvish', 250], ['John', 495],
          ['Sai', 400], ['Eric', 350], ['Adam', 450]]
```

```python
# Cols: Name, Univ_marks; there are many ways to do this
columns = ['Name', 'Univ_marks']
df = pd.DataFrame(values, columns=columns)
print(df)
```

```python
df['Grade_percent'] = (df.Univ_marks / 500) * 100
```

```python
print(df)
```

```python
# inserting a column at a specific location in the table
age = [20, 21, 19, 25, 33, 99]
df.insert(1, 'Age', age)
print(df)
```

```python
major = ['Art', 'Language', 'Maths', 'Science', 'Nursing', 'Business']
df.insert(2, 'Dept', major)
print(df)
```

```python
address = ['MI', 'GA', 'FL', 'NH', 'RI', 'MA']
df.assign(address=address)
```

```python
df1 = df.assign(address=address)
print(df1)
```
