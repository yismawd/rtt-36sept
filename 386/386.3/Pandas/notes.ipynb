#### Pandas DataTypes
-   series (`s`) - single dimensional array. Single column.
    -   homogenous - 1 data type per series
-   dataframe (`df`) - two dimensional array (table). Collection of columns.

```python
# declare series
ser = pd.Series(list('abcdefghijklmnopqrstuvwxyz'))

# Give it a name attribute
ser.name = 'alphabet'

# .head(n) - shows top 5 records - n = number of records
print(ser.head())

# .tail(n) - shop the bottom 5 records - n = number of records
print(ser.tail())
```

```python
# Declare a series from a dictionary
d = {
    'a': 0.0,
    'b': 1.0,
    'c': 2.0
}

# For a dictionary becoming a series, the keys become the indexes
float_series = pd.Series(d)

print(float_series)
```

```python
# Index is specified, corresponding data will be pulled into the series
customer_index = pd.Series(d, index=['a', 'y', 'x'])

print(customer_index)
```

```python
calories = {
    'day1': 2200,
    'day2': 2500,
    'day3': 2700,
}


index = ['day1', 'day2']
# Create a series with the following(or similare dictionary). but only with days 1 and 2
calories_series = pd.Series(calories, index=index)

print(calories_series)
```

```python
# Series of scalar value. Each index will have the same value
scalar_series = pd.Series(5.0, index=['a', 'b', 'c', 'd', 'e'])

print(scalar_series)
```

```python
# Element-wise arithmatic with series
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

# Just like arrays, math with series, are element-wise
print(ser1 + ser2)
print(ser1 - ser2)
print(ser1 * ser2)
print(ser1 / ser2)
```

```python
# Compare series - element-wise
# Element-wise arithmatic with series
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

print(ser1 > ser2)
```

```python
# Slicing a DataFrame
- pull out specific parts of a dataframe 
- to examin a subset, remove unwanted / unneeded data
data isolation, removal of clutter
- 'df[[cols1, col2, col3, col4]]' - slicing using double square bracket synthax 
```

```python
# Initializing the nested list with Data set
employee_list = [['James', 36, 75, 5428000],
               ['Villers', 38, 74, 3428000],
               ['VKole', 31, 70, 8428000],
               ['Smith', 34, 80, 4428000],
               ['Gayle', 40, 100, 4528000],
               ['Rooter', 33, 72, 7028000],
               ['Peterson', 42, 85, 2528000],
               ['John', 41, 85, 1528000],

]
columns = ['Name', 'Age', 'Weight (KGs)', 'Salary']
df = pd.DataFrame(employee_list, columns=columns)
print(df.head())
```

```python
# slicing Functions
- 'df.loc[]' - label based location (column names to locate and slice)
- 'df.iloc[start:stop:step (rows) , start:stop:step (cols)]' - index based location (index numbers to locate and slice)
- with the python indexing operator. STOP IS ALWAYS NON_INCLUSIVE
```

```python
# python indexing operator
emp_subset = df.iloc[ : , 0:2]
print(emp_subset)
```

```python
# using iloc, how can you grab the first 2 records of the first 2 columns
emp_sub_subset = df.iloc[0:2, 0:2]
print(emp_sub_subset)
```

```python
# use a list in the column argument to specificy specific lists
print(df.iloc[:, [0, 2, 3]])
```
