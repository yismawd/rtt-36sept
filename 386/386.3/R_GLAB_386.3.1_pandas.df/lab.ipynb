```python
# R_GLAB_386.3.2_pandas_df
import numpy as np
import pandas as pd
```

```python
# dictionary to make into frame
d = {
    'x':[1, 2, 3],
    'y':[2, 4, 8],
    'z':100  # scalar values, repeated for all records
}

print(pd.DataFrame(d))
```

```python
# use the index arguemnet to creat custom index values (index parameter)
# use columns argument/parameter to select columns or give names if none present
print(pd.DataFrame(d, index=[100, 200, 300], columns=['x', 'y', 'a']))
```

```python
student_dict = {
    'Name': ['Joe', 'Nat', 'Harry'],
    'Age': [20, 21, 28],
    'Marks': [85.10, 77.80, 91.54]
}
student_df = pd.DataFrame(student_dict)
print(student_df)
```

```python
# set_index() -> allows you to set the index
# set the index as one of the column
student_df = student_df.set_index('Name')
print(student_df)
```

```python
data = {
    'Product': ['Laptop', 'Tablet', 'Phone'],
    'Price': [1200, 300, 800],
    'Quantity': [50, 150, 100]
}
df = pd.DataFrame(data)
print(df)
```

```python
# accessing index attribute and using rangeindex to creat customer index range. Range index can also improve optimization, see docs for details
df.index = pd.RangeIndex(start=1, stop=4, step=1)
print(df)
```

```python
patientID = [101, 102, 103, 104]
name = ['Daniel', 'John', 'Abbi', 'Charlie']
dob = ['2023-01-01', '2024-01-02', '3/10/1987 14045', '13th of October, 2026']

# zip() - function combines lists to make a tuple of tuples
# tuples are an iterator, which is what the DF function takes
pt_df = pd.DataFrame(zip(patientID, name, dob))
print(pt_df)
```

```python
pt_df = pd.DataFrame(zip(patientID, name, dob), columns=['patient_id', 'Name', 'date_of_birth'])
print(pt_df)
```

```python
stocks = ['IBM', 'APPLE', 'TWTTER', 'GE', 'MSFT']
prices = [115.00, 119.14, 19.77, 25.99, 26]
pt_df = pd.DataFrame(zip(stocks, prices), columns=['stocks', 'prices'])
print(pt_df)
```

```python
data = [
    {'x': 1, 'y': 2, 'z': 3},
    {'x': 2, 'y': 10, 'z': 15},
    {'x': 4, 'y': 13, 'z': 16}
]

print(pd.DataFrame(data))
```

```python
data = [
    [1, 2, 3],
    [2, 4, 100],
    [3, 8, 100]
]

print(pd.DataFrame(data, columns=['a', 'b', 'c']))
```

```python
arr = np.array([[1, 2, 3], [2, 10, 15], [4, 8, 16]])

print(pd.DataFrame(arr, columns=['x', 'y', 'z']))
```

```python
arr = np.array([[1, 2, 3], [2, 10, 15], [4, 8, 16]])

df = pd.DataFrame(arr, columns=['x', 'y', 'z'], copy=False)
```
