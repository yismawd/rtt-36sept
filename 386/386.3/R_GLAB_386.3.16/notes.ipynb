```python
import pandas as pd
import numpy as np
```

```python
dateRange = pd.date_range(start = "07-01-2023", end = "07-31-2025")
dateRange
```

```python
dateRange = pd.date_range(start="01-01-2023", end="07-31-2025", freq="6min")
dateRange
```

```python
dateRange = pd.date_range(start="2023-07-01", end="2025-07-31", freq="8min")
print(dateRange)
```
