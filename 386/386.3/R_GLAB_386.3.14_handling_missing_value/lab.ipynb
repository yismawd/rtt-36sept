# initial data integrity checklist - steps to follow when you get new data setup
- look at shape (rows, columns) - let you know how many variables you are working with and how larde the dataset is
- look at the top 2 and bottom 5 records (.head(), tail())
- look at the data types (.info(), .dtypes())
- look atdescriptive statistics (.describe()) - making sure every falls in normal range, taking note if not
- look at values counts of categorical variables (.value_counts), (unique())
- take note of null/missing values (df.isnull().sum())
- examine for duplicate records (df.duplicated().sum())

# what to do when you find missing data?
- take note of what columns are missing data
- why is it missing? -mnot always possible, but you should take a look anyway
- decided what to do with missing data:
   - drop missing records ( 'df.dropna()' ) - not recommended you might lose value data
   - impute (replace) missing values with a placeholder
   - if order data forward or back fill day
   -  categorical: impute those values with mode value
   - quantitative data: impute with mean (normal distribution), or Median (for skewed or non-normal distribution)

```python
df = df.fillna('None')
```

```python
print(df.describe())
```

```python
headers = ['Name', 'Age', 'Weigh (kg)', 'Salary']

df = pd.DataFrame(emp_list, columns=headers)
```

```python
print('-'*10, 'Before Handling Missing Values', '-'*10)
print(df)
```

```python
median_Weight = df['Weigh (kg)'].median()
print(median_Weight)
```

```python
# Fill missing values in the existing "Weigh (kg)" column using the median
df['Weigh (kg)'] = df['Weigh (kg)'].fillna(median_Weight)

# Optional: verify the result
print(df)
```

```python
mean_salary = df['Salary'].mean()
print(f"Mean Salary: {mean_salary}")
```

```python
# Fill remaining missing Salary values using the mean salary already computed
df['Salary'] = df['Salary'].fillna(mean_salary)

print(df)
```

```python

headers = ['Name', 'Age', 'Weigh (kg)', 'Salary']

df = pd.DataFrame(emp_list, columns=headers)

print('-'*10, 'Before Handling Missing Values', '-'*10)
print(df)
```

```python
subset = df[['Weigh (kg)', 'Salary']].fillna('Pending')
print(subset)
```

```python
mean_salary = df['Salary'].mean()
median_weight = df['Weigh (kg)'].median()

df2 = df.fillna(value={'Name': 'Unknown', 'Salary': mean_salary, 'Weigh (kg)': median_weight})

print(df2)
```

```python
technologies = {
    'Courses':["Spark",np.nan,"Hadoop","Python","pandas",np.nan,"Java"],
    'Fee' :[20000,25000, np.nan,22000,24000,np.nan,22000],
    'Duration':[np.nan,'40days','35days', np.nan,'60days','50days','55days'],
    'Discount':[1000,np.nan,1500,np.nan,2500,2100,np.nan]
              }
df = pd.DataFrame(technologies)
print(df)
```
