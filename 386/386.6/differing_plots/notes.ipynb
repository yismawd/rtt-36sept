Scatter plot - used to visualize relationship between two or more variables
- plt.scatter(x, y, marker, alpha, color, cmap, s=size)

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib 
```

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
```

```python
plt.grid(alpha=0.5, linestyle='--')
plt.scatter(x, y, color='red', marker='*', s=90)
plt.show()
```

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([20, 50, 100, 200, 500, 1000, 60, 90, 10, 300, 600, 800, 75])
plt.figure(figsize=(8, 3))
plt.grid(alpha=0.5, linestyle='--')
plt.scatter(x, y, s=sizes, color='red', alpha=(sizes / sizes.max()))
```

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([20, 50, 100, 200, 500, 1000, 60, 90, 10, 300, 600, 800, 75])
fig, axes = plt.subplots(1, 1, figsize=(8, 3))
axes.grid(alpha=0.5, linestyle='--')
axes.scatter(x, y, s=sizes, color='red', alpha=(sizes / sizes.max()))   
```

```python
x = np.random.randint(100, size=(100))
y = np.random.randint(100, size=(100))
colors = np.random.randint(100, size=(100))
sizes = np.random.randint(100, size=(100))

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='winter') # you can change the colormap, search it on line
plt.title('Chart Title')
plt.xlabel('x-axis')
plt.ylabel('y-axis')

plt.colorbar(label='Color value') # color legend()
plt.gca().set_facecolor('lightgray')
plt.grid(alpha=0.8, linestyle='--')
plt.show()
```

```python
agegroup = range(40, 70)
purchase_data = np.abs(np.random.rand(30) * 20)
plt.scatter(agegroup, purchase_data, s=50, alpha=0.8)
plt.scatter(agegroup, purchase_data)
plt.xlabel('Age Group')
plt.ylabel('Purchase Data')
plt.title('Scatter Plot of Age vs Purchase Data')
plt.grid(alpha=0.8, linestyle='--')

agegroup2 = range(20, 40) # age is <= 39, age groups can overlap but the range should be the same size
purchase_data2 = np.abs(np.random.rand(20) * 40)
plt.scatter(agegroup2, purchase_data2, s=40, alpha=0.8)
plt.show()
```

```python
# Barcharts - categorical comparisons
- bar chart, histograms (quantitative), horizontal barchars, stacked barcharts, clustered column/barchart
```

```python
courses = ['Machine Learning', 'SQL Basics', 'Python Pro', 'Data Analytics', 'Cloud Databases']
num_of_students = [50, 90, 30, 130, 16]
plt.figure(figsize=(6, 2))
plt.bar(courses, num_of_students, color='green', width=0.5)
plt.title('Number of Student Per Class')
plt.xlabel('Class Name')
plt.ylabel('Number of Students')
plt.show()
```

```python
plt.figure(figsize=(6, 2))
plt.barh(courses, num_of_students, color='green', height=0.5)
plt.title('Number of Student Per Class')
plt.ylabel('Class Name')
plt.xlabel('Number of Students')
plt.show()
```

```python
courses = ['Machine Learning', 'SQL Basics', 'Python Pro', 'Data Analytics', 'Cloud Databases']
course_group1 = [50, 90, 30, 130, 16]
course_group2 = [25, 60, 100, 12, 50]
plt.figure(figsize=(6, 2))
plt.bar(courses, course_group1, width=0.3, label='Group 1')

plt.bar(courses, course_group2, width=0.3, bottom=course_group1, label='Group 2')

plt.title('Number of Student Per Class')
plt.xlabel('Class Name')
plt.ylabel('Number of Students')
plt.legend()

plt.show()
```

```python
years = ['2016', '2017', '2018', '2019']
rice_price = [57, 50, 47, 30]
wheat_price = [43, 50, 53, 70]


plt.figure(figsize=(6, 2))
plt.bar(years, rice_price, width=0.3, label='Rice')

plt.bar(years, wheat_price, bottom=rice_price, width=0.3, label='Wheat')


plt.title('The prefrence of Rice vs Wheat')
plt.xlabel('Years')
plt.ylabel('Preferences %')
plt.legend()

plt.show()
```

```python
years = ['2016', '2017', '2018', '2019']
rice =  np.array([57, 50, 47, 30])
wheat =  np.array([43, 50, 53, 70])
mustard =  np.array([10, 16, 13, 15])

plt.figure(figsize=(6, 3))
plt.bar(years, rice, width=0.3, label='Rice')
plt.bar(years, mustard, bottom=rice + wheat, width=0.3, label='mustard')
plt.bar(years, wheat, bottom=rice, width=0.3, label='Wheat')

plt.title('The production of Rice, Wheat and Mustard')
plt.xlabel('Years')
plt.ylabel('Quantity')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))

plt.show()

```
