```python
import numpy as np
```

```python
python_list = [1, 2, 3, 4, 5]

numpy_1d_array = np.array(python_list)
print(numpy_1d_array)
```

```python
# attributes
# print the array
# print the shape of the array
# print the size of the array
numpy_1d = numpy_1d_array
print('Array: ', numpy_1d)
print('Shape: ', numpy_1d.shape)
print('Size: ', numpy_1d.size)
print('DataType: ', numpy_1d.dtype)
```

```python
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 3, 6, 8])
# element wise addition and save into new variable
# element wise subtraction and save into new variable
# element wise multiplication and save into new variable
# element wise division and save into new variable
# print out variables
print('Array 1: ', arr1)
print('Array 2: ', arr2)

print('-' * 30)

addition = arr1 + arr2
subtraction = arr1 - arr2
multiplication = arr1 * arr2
division = arr1 / arr2

print('Addition: ', addition)
print('Subtraction: ', subtraction)
print('Multiplication: ', multiplication)
print('Division: ', division)
```

```python
data = np.array([85, 90, 78, 92, 88, 85, 95])

# Calculate the mean - save in variables
mean = np.mean(data)
# Calculate the median - save in variables
median = np.median(data)
# Calculate the standard deviation - save in variables
stDev = np.std(data)
# Calculate the range - save in variables
range = np.max(data) - np.min(data)
# Calculate the variance - save in variables
variance = np.var(data)

#Print out your values
print(f'Mean: {mean:.2f}')
print('Median: ', median)
print('StDev: ', stDev)
print('Range: ', range)
print('Variace: ', variance)
```
