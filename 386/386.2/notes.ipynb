```python
import numpy as np

print('Numpy Version: ', np.__version__)
```

```python

# single dimensional arrays (Vector)
oneD = np.array(
    [1, 2, 3]
)

print(oneD)

# Two dimensional array (Matrix) Grid. Cery to a table
twoD = np.array(
    [
        [1, 2, 3],
        [3, 4, 6],
        [7, 8, 9]
    ]
)

print(twoD)

# Three dimensional Array (Tensor). set of matrices
threeD = np.array(
    [
        [
            [1, 2],
            [3, 4]
        ],
        [
            [5, 6],
            [7, 8]
        ]
    ]
)

print(threeD)
```

```python
## Declaring arrays in python
- '.array() - declaring simple arrays with lists, tuples, ranges
- '.arange() - range function to create an array
    - syntax np.arange(start, stop, step, dtype)
    
```

```python
one_to_100 = np.arange(100)
print(one_to_100)
# array from 1-30 not including 30
one_thirty = np.arange(1, 30)
print(one_thirty)
# array from 10-20 by 2
ten_twenty = np.arange(10, 20, 2)
print(ten_twenty)
# array from -3-10 by 4, verify dtype is int
print(np.arange(-3, 10, 4, dtype=int))
```

```python
# numpy array attributes 
- 'array.shape' - tuple with the num of rows and columns
- 'array.size' - total number of elements (records) in array
- 'array.dtype' - data type of arrays
- 'array.ndim' - number of dimentions
```

```python
print(oneD)
print(oneD.shape)
print(oneD.size)
print(oneD.dtype)
print(oneD.ndim)
```

```python

arr = np.array([1, 2, 3, 4, 5])

element = arr[3]
print('Element: ', element)

slice = arr[1:4]
print('Slice: ', slice)

print(arr)
```

```python
# Arithmetic and NumPy Arrays
# Arithmetic with arrays and scalar values compute element-wise (PER ELEMENT, for each element)
arr = np.array([1, 2, 3])
arr2 = np.array([9, 8, 7])

print(arr + 1) #[2, 3, 4]
print(arr * 3) #[3, 6, 9]
print(arr / 2) # [0.5, 1.0, 1.5]

print(arr + arr2) # [10, 10, 10] added elements at the same index together.
```
