```python

# single dimensional arrays (Vector)
oneD = np.array(
    [1, 2, 3]
)

print(oneD)

# Two dimensional array (Matrix) Grid. Cery to a table
twoD = np.array(
    [
        [1, 2, 3],
        [3, 4, 6],
        [7, 8, 9]
    ]
)

print(twoD)

# Three dimensional Array (Tensor). set of matrices
threeD = np.array(
    [
        [
            [1, 2],
            [3, 4]
        ],
        [
            [5, 6],
            [7, 8]
        ]
    ]
)

print(threeD)
```

```python
## Declaring arrays in python
- '.array() - declaring simple arrays with lists, tuples, ranges
- '.arange() - range function to create an array
    - syntax np.arange(start, stop, step, dtype)
    
```

```python
one_to_100 = np.arange(100)
print(one_to_100)
# array from 1-30 not including 30
one_thirty = np.arange(1, 30)
print(one_thirty)
# array from 10-20 by 2
ten_twenty = np.arange(10, 20, 2)
print(ten_twenty)
# array from -3-10 by 4, verify dtype is int
print(np.arange(-3, 10, 4, dtype=int))
```

```python
# numpy array attributes 
- 'array.shape' - tuple with the num of rows and columns
- 'array.size' - total number of elements (records) in array
- 'array.dtype' - data type of arrays
- 'array.ndim' - number of dimentions
```

```python
print(oneD)
print(oneD.shape)
print(oneD.size)
print(oneD.dtype)
print(oneD.ndim)
```

```python

arr = np.array([1, 2, 3, 4, 5])

element = arr[3]
print('Element: ', element)

slice = arr[1:4]
print('Slice: ', slice)

print(arr)
```

```python
# Arithmetic and NumPy Arrays
# Arithmetic with arrays and scalar values compute element-wise (PER ELEMENT, for each element)
arr = np.array([1, 2, 3])
arr2 = np.array([9, 8, 7])

print(arr + 1) #[2, 3, 4]
print(arr * 3) #[3, 6, 9]
print(arr / 2) # [0.5, 1.0, 1.5]

print(arr + arr2) # [10, 10, 10] added elements at the same index together.
```

```python
import numpy as np

# NumPy Aggregations
arr = np.array([1, 2, 3, 4, 5])

sum = np.sum(arr)
mean = np.mean(arr)
median = np.median(arr)
stDev = np.std(arr)
min = np.min(arr)
max = np.max(arr)

print('Sum: ', sum)
print('Avg: ', mean)
print('Median: ', median)
print('Standard Deviation: ', stDev)
print('Min: ', min)
print('Max: ', max)
```

```python
# transpose or reshape
matrix = np.array(
    (
        [1, 2, 3],
        [4, 5, 6]
    )
)
print('before transpose')
print(matrix)
#transpose, pivot table 90 degrees
transpose_matrix = matrix.T
print('\nTransposed Matrix')
print(transpose_matrix)

reshape_matrix = matrix.reshape(3, 2)
print('\nreshaped Matrix')
print(reshape_matrix)
```

```python
# logical Operators AND, OR NOT
a = np.array([True, True, False, False])
b = np.array([True, True, False, False])
print(np.logical_or(a, b))

print('\nLogical Or') # compare each element in a list
print(np.logical_or(a, b))
```

```python
c = np.array([True, True, False, False])
print(np.logical_not(c)) # flips the boolean value
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([1, 2, 3, 4])
c = np.array([2, 3, 4, 5])

print(np.array_equal(a, b)) # compares the entire array looking for difference
print(np.array_equal(a, c))
```

```python
# sorting arrays
a = np.array([3, 13, 6, 23, 14, 100, -1, -8, 32])
print(np.sort(a))
```

```python
# searchin # use python membership operators
print(12 not in a)
print(3 in a)
```

```python
b = np.array([3, 13, 6, 23, 14, 100, -1, -8, 32])
print(np.argmin(b)) # return the index number of the lower/minimum value
print(np.argmax(b)) # return the index number of the max value
```

```python
# count_nonzero() - counts non-zero elements
a = np.array([0, 1, 2, 0, 13, 15, 0, 6])
print(np.count_nonzero(a))
b = np.array([
    [1, 2, 0],
    [0, 0, 0]
])
print(np.count_nonzero(b))
```
