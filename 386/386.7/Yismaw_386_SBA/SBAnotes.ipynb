# 7. Read the total profit of all months and show it using the Bar plot.

```python
plt.bar('month_number', 'total_profit', data=df, width=0.5)
plt.xlabel('Month number')
plt.xticks(range(0, 13, 2))
plt.ylabel('Total profit')
plt.yticks([100000, 200000, 300000, 400000, 500000])
plt.title('Company profit per month')
plt.grid(True, alpha=0.8, linestyle='--')

plt.show()
```

# 8. Read the total profit of all of the months, and show the line plot 

```python
plt.figure()
plt.plot(
    df['month_number'],
    df['total_profit'],
    linestyle=':',
    color='red',
    linewidth=3,
    marker='o',
    markerfacecolor='black',
    markeredgecolor='black',
    label='Profit data of last year'
)

plt.xlabel('Month number')
plt.xticks(range(0, 13))
plt.ylabel('Total profit')
plt.yticks([100000, 200000, 300000, 400000, 500000])
plt.title('Company profit per month')
plt.legend(loc='lower right')
plt.show()
```

# 9. Print all of the product sales data and show it using a multi-line plot.

```python
product_columns = [
    'facecream', 'facewash', 'toothpaste',
    'bathingsoap', 'shampoo', 'moisturizer'
]

plt.figure()
for product in product_columns:
    plt.plot(df['month_number'], df[product], marker='o', label=product.title())

plt.xlabel('Month Number')
plt.ylabel('Sales units in numbers')
plt.yticks([1000, 2000, 4000, 6000, 8000, 10000, 12000, 15000, 18000])
plt.title('Sales Data')
plt.xticks(df['month_number'])
plt.legend(['Facecream Sales Data', 'Facewash Sales Data', 'Toothpaste Sales Data',
            'Bathingsoap Sales Data', 'Shampoo Sales Data', 'Moisturizer Sales Data'])
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

# 10. Read “bathingsoap” sales data for each month and show it using a scatter plot.

```python
plt.figure()
plt.scatter(df['month_number'], df['bathingsoap'])

plt.xlabel('Month Number')
plt.ylabel('Number of units Sold')
plt.title('bathing soap Sales data')
plt.xticks(df['month_number'])
plt.yticks([ 6000, 8000, 10000, 12000, 14000])
plt.grid(True, linestyle='--')
plt.legend(['bathing soap Sales data'], loc='upper left')
plt.show()
```
