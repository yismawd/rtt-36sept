```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B','A','B'],
    'Value': [10, 15, 20, 25, 30, 35,40,50],
    'Fee' :[20000,25000,26000,22000,24000,35000,4000,5600],
    'Duration':['30day','40days','35days','40days','60days','60days','50days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000,500,600]
}
```

```python
df = pd.DataFrame(data)
print(df.head())
```

```python
# calculate the total sum of the Value column using the aggregate function
total_sum = df['Value'].agg('sum')
print(total_sum)
```

```python
total_sum1 = df[['Value']].agg('sum')
print(f'total sum: {total_sum}')
```

```python
df['Value'].agg('cumsum')
```

```python
df[['Value']].agg('cumsum')
```

```python
# use both the agg and alternative to find: mean, max, count
```

```python
# Using agg
value_stats_agg = df['Value'].agg(['mean', 'max', 'count'])
print("Using agg:")
print(value_stats_agg)

# Alternative using built-in methods
value_stats_alt = {
    'mean': df['Value'].mean(),
    'max': df['Value'].max(),
    'count': df['Value'].count()
}
print("Alternative:")
print(value_stats_alt)
```

```python
# OR
mean1 = df['Value'].agg('mean')
print(f'mean: {mean1}')
mean2 = df[['Value']].mean()
print(f'mean: {mean2}')
```

```python
max1 = df['Value'].agg('max')
print(f'max: {max1}')
max2 = df[['Value']].max()
print(f'max: {max2}')
```

```python
count1 = df['Value'].agg('count')
print(count1)

count2 = df[['Value']].count()
print(count2)
```

Multiple Aggregations

```python
result = df[['Fee', 'Discount']].agg('sum')
result
```

```python
result = df[['Fee', 'Discount']].sum()
result
```

```python
result = df[['Fee', 'Discount']].agg(['min', 'max', 'mean'])
result
```

```python
result_group = df.groupby('Category')[['Value', 'Fee', 'Discount']].agg(['sum', 'mean', 'min'])
result_group
```
