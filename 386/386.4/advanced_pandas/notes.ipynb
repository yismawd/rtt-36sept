```python
# handling multiple dataframes
- 'df.merge()' - similar to join in sql - columns or keys/columns; make the data set wider, add column
- 'df.join()' - similar to a join in sql - combinnes dataframes on an index, make the data set wider, add columns
- 'df.concat' - similar to a union in sql - add rows, stacking DF on top of each other, add rows and make the dataset longer
- 'df.append()' - deprecated, don't use, use concat instead
```

```python
left_df = pd.DataFrame({'Courses': ["Spark","PySpark","Python","pandas","Java"],
                    'Fee' : [20000,25000,30000,24000,40000],
                    'Duration':['30day','40days','60days','55days','50days']})

right_df = pd.DataFrame({'Courses': ["Java","PySpark","Python","pandas","Hyperion","html"],
                    'Fee': [20000,25000,30000,24000,40000,4000],
                    'Percentage':['10%','20%','25%','20%','10%','50%']})
```

```python
# Merge the two DataFrames on the shared "Courses" column
merged_df = left_df.merge(
    right_df,
    on="Courses",
    how="inner",
    suffixes=("_left", "_right")
)
print("Inner merge:")
print(merged_df)

# Left join to keep all rows from left_df
left_join_df = left_df.merge(
    right_df,
    on="Courses",
    how="left",
    suffixes=("_left", "_right")
)
print("\nLeft join:")
print(left_join_df)
```

## JOIN

```python
technologies = {
    'Courses':["Spark","PySpark","Python","pandas"],
    'Fee' :[20000,25000,22000,30000],
    'Duration':['30days','40days','35days','50days'],
              }
index_labels=['r1','r2','r3','r4']
df1 = pd.DataFrame(technologies,index=index_labels)

print("First DataFrame:\n", df1)

technologies2 = {
    'Courses':["Spark","Java","Python","Go"],
    'Discount':[2000,2300,1200,2000]
              }
index_labels2=['r1','r6','r3','r5']
df2 = pd.DataFrame(technologies2,index=index_labels2)

print("Second DataFRame:\n", df2)
```

```python
# df.join(other, lsuffix="_called", rsuffix="_other")
# by default it joins using what (=index values)
df1.join(df2, lsuffix='_left', rsuffix='_right')
```

```python
df1.join(df2, lsuffix='_left', rsuffix='_right', how='right')
```

```python
df1.join(df2, lsuffix='_left', rsuffix='_right', how='inner')
```

```python
df1.set_index('Courses').join(df2.set_index('Courses'), lsuffix='_left', rsuffix='_right')
```
