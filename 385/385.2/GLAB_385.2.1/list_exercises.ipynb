#references
- creating a copy of a list doesn't create a new list. it points to the same place in memory
- google: press by value vs pass vs reference
- changing one referenced list, will also change the other


```python
og_list=[1,2,3,4,5]
new_list=og_list
print('OG', og_list)
print('new', new_list)

new_list.append(6)
print('OG', og_list)
print('new', new_list)
```

# deep copy - a completely separate copy of a non primitive data type, not just a reference to the same place in memory
# shallow copy - a small separete copy of non primitive data type. 1st dimentsion is separete places in memory

```python

second_copy=og_list.copy()
print(og_list)
og_list.append(7)
print(og_list)
print(second_copy)
```

```python
# reverse, extends, sort, and clear methods
num_list=[1,2,3,4,5]
print(num_list)
num_list.reverse()
print(num_list)
num_list.extend([6,7,8,9])
print(num_list)
num_list.sort()
print(num_list)
```

```python
print(1 in num_list)
print(1 not in num_list)
```
