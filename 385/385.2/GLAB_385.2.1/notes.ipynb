### string 
collection of single characters
data access within lists and string is IDENTICAL
index numbers to access values



```python
greet = 'hello, world!'
print(greet[0])
print(greet[1])
print(greet[2])
```

python indexing operator
- can be used with strings and lists
- [ start : stop: step]
- by default, if we dont include the step, its 1


```python
print(greet[ : :-1])
print(greet[0:5])
```

type () function
- tells us what data type a variable is
- type(value) -> data type

```python
print(type(greet))
print(type(13))
print(type(True))
print(type(13.2))
```

typecasting functions (tochange the datatype of)
- str() - turns anything into a string
- float() - turns value into a float if possible
- int() - turns value into a int if possible

```python
age = '30'
print(type(age))
# if you permanently want to change a variable you have to re-assign it
age=int(age) 
print(type(age))
```
