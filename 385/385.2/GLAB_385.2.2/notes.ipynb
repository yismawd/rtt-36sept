# File Processing

#### File Paths:
-   Absolute Paths - Beings at the root folder. Starts at the base of your computer.
    -   ex: `/Volumes/ExtHD/PerScholas/RTT-36/385/385.3/notes/notes.ipynb`
-   Relative paths - Path is relative to the current (working) directory
    -   ex: `./notes.ipynb`

#### File Path Symbols
-   `/` - divider between folders/files
-   `.` - currect directory
-   `..` - One directory up the chain

#### Opening Files
-   `open(file_path, mode)` - function

#### File modes
-   `r` - read - if file does not exists (FileNotFoundError)
-   `w` - write - if file doesnt exist, it creates a new file
-   `a` - append - if file doesnt exist, it creates a new file
-   `x` - exlusive creation - if file doesnt exist, it creates a new file
-   `b` - binary
-   `t` - text
-   `+` - read/write

```python
f = open('data/my_file.txt', mode='r') # opens the file

print(f.read()) # reads the file

f.close() # Always close your file to prevent extra resources and corruption of files.
```

#### Error Handling w/ Try: Except: Finally
-   `try:` - this block is code we want to run
-   `except:` - raise exceptions for errors we run into. Code to run if we get an error
-   `finally:` - code we run no matter what happens. errors or not

```python
try:
    f = open('data/my_file.txt', mode='r')

    print(f.read())

except Exception as e: # if an error, put in variable 'e' and print error to logs (terminal)
    print(f'❌ Error: {e}')

finally:
    f.close()
```

```python
# if opening file with 'with', all code goes into the with code block and the file auto closes at the end of the block
with open('data/my_file.txt', 'r') as f:
    print(f.read())
```

```python
def write_to_file(str, file_path):
    try:
        with open(file_path, 'w') as f:
            f.write(str + '\n')
            
        print('Success')
    except Exception as e:
        print(f'❌ Error: {e}')
```

```python
write_to_file('Dylan was here', 'data/my_file.txt')
```

```python
write_to_file('Darryl was here', 'data/my_file.txt')
```

```python
def append_to_file(str, file_path):
    try:
        with open(file_path, 'a') as f:
            f.write(str + '\n')
            
        print('Success')
    except Exception as e:
        print(f'❌ Error: {e}')
```

```python
append_to_file('Darryl was here', 'data/my_file.txt')
```

# Comma Separated Values (CSV)

-   text file that uses structuring to arrange/represent tabular data.
-   Programs that handle large amounts of data, create/export csv files.
-   Simple, straightforward to workwith
-   Portable. interoperable
-   Pandas/NumPy are built to work directly with tabular data

```python
# Python built in module for handling csv's
import csv
```

```python
try:
    with open('data/country.csv', 'r', encoding='UTF8') as f:
        csv_res = csv.reader(f)

        # Enumerate the iterable (gives us key, value pair tuples with the index no.)
        # destructured the tuple into index and data
        for line_no, line in enumerate(csv_res):
            if line_no == 0:
                print('Header:')
                print(line)
                print('Data:')
            else:
                print(line)

except Exception as e:
    print(f'❌ Error: {e}')
```

```python
try:
    with open('data/country.csv', 'r', encoding='UTF8') as f:
        csv_res = csv.reader(f)

        next(csv_res) # skip first row of an iterable

        for line in csv_res:
            print(line)
except Exception as e:
    print(f'❌ Error: {e}')
```

```python
try:

    fieldnames = ["Name", "total_area", "cc1", "cc3"]

    with open('data/country.csv', 'r', encoding='UTF8') as f:
        csv_res = csv.DictReader(f, fieldnames=fieldnames)

        next(csv_res)

        for line in csv_res:
            print(line)
            # print(f'{line['name']} has an area of {line['area']} km**2')

except Exception as e:
    print(f'❌ Error: {e}')
```

#### JSON JavaScript Object Notation
-   industry standard notation, for nested/heirarchical data.
-   NoSQL (Not Only SQL)
-   Looks like a list of dictionaries.
-   `[{}, {}, {}]`
-   Flexible schema information
    -   Allows for the representation of complex data structures
-   serialization - turning json object into string for transport
    -   `json.dump()` - serializes python object in file/stream
    -   `json.dumps()` - serialize python object into string
-   deserialization - turing string back into json object for use
    -   `json.load()` - parse json file (not string) into a python dictionary
    -   `json.loads()` - parse json string into a python dictionary

```python
# JSON module, build in python module used for handling json files
import json
```

```python
# Opening json file in read mode
try:
    with open('data/data.json', 'r') as f:
        # using json.load() to deserialize json data
        data = json.load(f)
        print(data)
except Exception as e:
    print(f'❌ Error: {e}')
```

```python
cat = {
    "name": "Bandit",
    "age": 9,
    "city": "East Greenwich"
}

cat1 = {
    "name": 'Thumbs',
    'age':6,
    'city': 'Tessas Kingdom'
}

cats = [cat, cat1]
# Turns dictionary into string for transport
json_string = json.dumps(cat)

print(type(json_string))
print(json_string)
```

```python
# I want to create/write, a new json file
try:
    # Both writing and appending, create a file if it doesnt exist
    with open('data/cats.json', 'a') as f:
        json.dump(cats, f)
        
except Exception as e:
    print(f'❌ Error: {e}')
```
