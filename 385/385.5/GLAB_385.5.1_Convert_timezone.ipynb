```python
def convert_tz(dt, from_tz, to_tz):
    #create a tz object for each tz
    from_tz = pytz.timezone(from_tz)
    to_tz = pytz.timezone(to_tz)
  
    # how to add tzinfo to an existing date time object
    # reassigning passed in dt with new timezone
    dt = dt.replace(tzinfo=from_tz)
    print(dt.astimezone(to_tz))
```
